```python
import matplotlib.pyplot as plt
```

```python
def malthus ( population :int, iterations : int ) -> int :
    historique = [population]
    for i in range(iterations) :
        population = 2 * population
        historique.append(population)
    return historique
```

```python
pop_depart = 50
nb_iterations = 5
populations = malthus(pop_depart, nb_iterations)
temps = list(range(len(populations)))
plt.figure(figsize=(9, 5))
plt.plot(temps, populations, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus simple")

plt.title("Évolution de la population - Modèle de Malthus", fontsize=14)
plt.xlabel("Temps (Itérations / n)", fontsize=12)
plt.ylabel("Taille de la population (yn)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```
