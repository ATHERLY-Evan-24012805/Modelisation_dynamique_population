```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import fsolve
import math
```

```python
def malthus ( population :int, iterations : int ) -> list :
    historique = [population]
    for i in range(iterations) :
        population = 2 * population
        historique.append(population)
    return historique
```

```python
pop_depart = 50
nb_iterations = 5
populations = malthus(pop_depart, nb_iterations)
temps = list(range(len(populations)))
plt.figure(figsize=(9, 5))
plt.plot(temps, populations, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus simple")

plt.title("Évolution de la population - Modèle de Malthus", fontsize=14)
plt.xlabel("Temps (Itérations / n)", fontsize=12)
plt.ylabel("Taille de la population (yn)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```

```python
def malthus_taux(y0 : float,a : float, t_max: float, nbPoints : int) -> list :
    """
    y0 : population initiale
    a : taux d'accroissement
    t_max : temps maximal
    nbPoints : nombre de points à tracer
    la formule est dans le cours page 26
    """
    t = np.linspace(0, t_max, nbPoints)
    y = y0 * math.e**(a*t)
    return t,y
```

```python
y0 = 10
a = 0.2
tmax = 20.0
temps,population = malthus_taux(y0,a,tmax,100)
print(malthus_taux(y0,a,tmax,100))
plt.figure(figsize=(9, 5))
plt.plot(temps, population, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus avec taux d'accroissement")

plt.title("Évolution de la population - Modèle de Malthus avec croissance exponentielle", fontsize=14)
plt.xlabel("Temps t", fontsize=12)
plt.ylabel("Taille de la population", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```

```python
def f(y, a):
    return y * a

def explicit_euler_method(y0, t0, h, t_max, a):
    t_values = np.arange(t0, t_max + h, h)
    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    for i in range(len(t_values) - 1):
        y_values[i + 1] = y_values[i] + h * f(y_values[i], a)

    return t_values, y_values
```

```python
y00 = 20
t0 = 0
h = 0.5
t_max = 20
a = 0.2

temps2 = explicit_euler_method(y00,t0,h,t_max,a)[0]
population2 = explicit_euler_method(y00,t0,h,t_max,a)[1]
plt.figure(figsize=(9, 5))
plt.plot(temps2, population2, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Méthode explicite d'euler")

plt.title("Évolution de la population - Méthode explicite d'euler", fontsize=14)
plt.xlabel("Temps t", fontsize=12)
plt.ylabel("Taille de la population", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()

print(explicit_euler_method(y00,t0,h,t_max,a))
```

```python
def implicit_euler_method(y0, t0, h, t_max, a):
    t_values = np.arange(t0, t_max + h, h) 
    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    for i in range(len(t_values) - 1):
        # équation à résoudre: y_values[i + 1] = y_values[i] + h * f(y_values[i+1], a)
        # y_values[i + 1] - y_values[i] - h * f(y_values[i], a) = 0
        # Z - y_values[i] - h * f(Z) = 0
        fonction_eq = lambda Z : Z - y_values[i] - h * f(Z, a)
        # On cherche la meilleure estimation de y[i + 1]; y[i] est la meilleure car le pas est faible et le valeurs ne change pas beaucoup
        point_initial = y_values[i] # fsolve résout par itération on définit ici le point de départ
        y_values[i + 1] = fsolve(fonction_eq, point_initial)
    return t_values, y_values
```

```python
y000 = 20
t00 = 0
h0 = 0.5
t_max0 = 10
a0 = 0.2

temps2 = implicit_euler_method(y000,t00,h0,t_max0,a0)[0]
population2 = implicit_euler_method(y000,t00,h0,t_max0,a0)[1]
plt.figure(figsize=(9, 5))
plt.plot(temps2, population2, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Méthode explicite d'euler")

plt.title("Évolution de la population - Méthode explicite d'euler", fontsize=14)
plt.xlabel("Temps t", fontsize=12)
plt.ylabel("Taille de la population", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()

print(implicit_euler_method(y00,t0,h,t_max,a))
```

```python
def calculer_aire_rectangle(t_values : list, y_values :list) -> int : 
    largeur = np.diff(t_values)
    hauteur = y_values
    aires = largeur * hauteur
    aire_totale = np.sum(aires)
    
    return aire_totale
```

```python
t_values = implicit_euler_method(y000,t00,h0,t_max0,a0)[0]
y_values = implicit_euler_method(y000,t00,h0,t_max0,a0)[0]
aire = calculer_aire_rectangle(t_values,y_values)
print(aire)
```
