## Malthus
Le modèle de Malthus est très simple, c'est un modele qui vient simuler la reproduction de bactéries ou a chaque instant une bactérie se divise en deux créant ainsi une courbe exponentielle.

```python
# Malthus 
# A chaque instant la population se multiplie par deux 
def malthus(population : int, iterations : int):
    h = [population]
    for i in range (iterations):
        population = population * 2
        print( f"tic : {i} - population : {population}")
        h.append(population)
    return h
        
malthus(2, 10)
```

```python
pop_depart = 50
nb_iterations = 5
populations = malthus(pop_depart, nb_iterations)
temps = list(range(len(populations)))
plt.figure(figsize=(9, 5))
plt.plot(temps, populations, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus simple")

plt.title("Évolution de la population - Modèle de Malthus", fontsize=14)
plt.xlabel("Temps (Itérations / n)", fontsize=12)
plt.ylabel("Taille de la population (yn)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```

```python
# Malthus_taux sans np
def malthus_taux_for (pop : int, taux_accroissement : float,iteration : int, nbPoints : int = 50 )-> list[float]:
    pas = iteration / (nbPoints-1)
    tab = [x for x in range (nbPoints)]
    for i in range(len(tab)) :
        # print(tab[i])
        tab[i] = tab[i] * pas
        
    
    result_tab = []
    for el in tab :
        # result_tab.append(pop * np.exp(a*el))
        result_tab.append(pop* 2.718281828459**(taux_accroissement*el))


    # print(t)
    print(tab)
    print(result_tab)
    return tab,result_tab


malthus_taux_for(10, 0.2, 20, 20)
```

```python
y0 = 10 # population
a = 0.2 # taux accroissement arbitraire
tmax = 20.0 # interation 
# temps,population = malthus_taux(y0,a,tmax)
temps,population = malthus_taux_for(y0,a,tmax,20)
print(f"temps : {temps}")
print(type(temps))
print(f"pop : {population}")
print(type(population))

plt.figure(figsize=(9, 5))
plt.plot(temps, population, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus avec taux d'accroissement")

plt.title("Évolution de la population - Modèle de Malthus avec croissance exponentielle", fontsize=14)
plt.xlabel("Temps (Itérations / n)", fontsize=12)
plt.ylabel("Taille de la population (yn)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```
