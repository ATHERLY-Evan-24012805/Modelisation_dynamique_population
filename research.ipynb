```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.optimize import fsolve
import math
```

```python
def malthus ( population :int, iterations : int ) -> list :
    historique = [population]
    for i in range(iterations) :
        population = 2 * population
        historique.append(population)
    return historique
```

```python
pop_depart = 50
nb_iterations = 5
populations = malthus(pop_depart, nb_iterations)
temps = list(range(len(populations)))
plt.figure(figsize=(9, 5))
plt.plot(temps, populations, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus simple")

plt.title("Évolution de la population - Modèle de Malthus", fontsize=14)
plt.xlabel("Temps (Itérations / n)", fontsize=12)
plt.ylabel("Taille de la population (yn)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()
```

```python
def malthus_taux (pop : int, taux_accroissement : float,iteration : int, nbPoints : int = 50 )-> list[float]:
    pas = iteration / (nbPoints-1)
    tab = [x for x in range (nbPoints)]
    for i in range(len(tab)) :
        # print(tab[i])
        tab[i] = tab[i] * pas
        
    
    result_tab = []
    for el in tab :
        # result_tab.append(pop * np.exp(a*el))
        result_tab.append(pop* math.e**(taux_accroissement*el))


    # print(t)
    # print(tab)
    # print(result_tab)
    return tab,result_tab
```

```python
y0 = 10
a = 0.2
tmax = 20.0
temps,population = malthus_taux(y0,a,tmax,20)
print(malthus_taux(y0,a,tmax,100))
plt.figure(figsize=(9, 5))
plt.plot(temps, population, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Modèle de Malthus avec taux d'accroissement")

plt.title("Évolution de la population - Modèle de Malthus avec croissance exponentielle", fontsize=14)
plt.xlabel("Temps t", fontsize=12)
plt.ylabel("Taille de la population", fontsize=12)
plt.legend()

plt.show()
```

```python
def f(y, a):
    return y * a

def explicit_euler_method(y0, t0, h, t_max, a):
    t_values = np.arange(t0, t_max + h, h)
    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    for i in range(len(t_values) - 1):
        y_values[i + 1] = y_values[i] + h * f(y_values[i], a)

    return t_values, y_values
```

```python
def implicit_euler_method(y0, t0, h, t_max, a):
    t_values = np.arange(t0, t_max + h, h) 
    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    for i in range(len(t_values) - 1):
        # équation à résoudre: y_values[i + 1] = y_values[i] + h * f(y_values[i+1], a)
        # y_values[i + 1] - y_values[i] - h * f(y_values[i], a) = 0
        # Z - y_values[i] - h * f(Z) = 0
        fonction_eq = lambda Z : Z - y_values[i] - h * f(Z, a)
        # On cherche la meilleure estimation de y[i + 1]; y[i] est la meilleure car le pas est faible et le valeurs ne change pas beaucoup
        point_initial = y_values[i] # fsolve résout par itération on définit ici le point de départ
        y_values[i + 1] = fsolve(fonction_eq, point_initial)
    return t_values, y_values
```

```python
y000 = 20
t00 = 0
h0 = 0.5
t_max0 = 10
a0 = 0.2

temps2 = implicit_euler_method(y000,t00,h0,t_max0,a0)[0]
population2 = implicit_euler_method(y000,t00,h0,t_max0,a0)[1]

temps3 = explicit_euler_method(y000,t00,h0,t_max0,a0)[0]
population3 = explicit_euler_method(y000,t00,h0,t_max0,a0)[1] 
plt.figure(figsize=(9, 5))
plt.plot(temps2, population2, marker='o', color='tab:blue', linestyle='-', linewidth=2, label=f"Méthode implicite d'euler")
plt.plot(temps3, population3, marker='s', color='tab:orange', linestyle='-', linewidth=2, label=f"Méthode explicite d'euler")

plt.title("Évolution de la population - Méthode explicite d'euler", fontsize=14)
plt.xlabel("Temps t", fontsize=12)
plt.ylabel("Taille de la population", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.show()

print(implicit_euler_method(y00,t0,h,t_max,a))
```

```python
def calculer_aire_rectangle(t_values : list, y_values :list) -> int : 
    largeur = np.diff(t_values)
    hauteur = np.diff(y_values)
    aires = largeur * hauteur
    aire_totale = np.sum(aires)
    
    return aire_totale
```

```python
def trapezes2(y0, t0, h, t_max, a):
    t_values = np.arange(t0, t_max + h, h)
    y_values = np.zeros(len(t_values))
    y_values[0] = y0
    for i in range(len(t_values) - 1):
        euler_explicit = y_values[i] + h * f(y_values[i], a)

        fonction_eq = lambda Z : Z - y_values[i] - h * f(Z, a)
        # On cherche la meilleure estimation de y[i + 1]; y[i] est la meilleure car le pas est faible et le valeurs ne change pas beaucoup
        point_initial = y_values[i] # fsolve résout par itération on définit ici le point de départ
        euler_implicite = fsolve(fonction_eq, point_initial)

        y_values[i+1] = (euler_explicit + euler_implicite)/2
    return t_values, y_values
```

```python
def verhulst(y, a, K):
    return a * y * (1 - (y / K))
```

```python
def explicit_euler_method_verhulst(y0, t0, h, t_max, a, K):
    t_values = np.arange(t0, t_max + h, h)
    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    for i in range(len(t_values) - 1):
        y_values[i + 1] = y_values[i] + h * verhulst(y_values[i], a,K)

    return t_values, y_values
```

```python
def implicit_euler_method_verhulst(y0, t0, h, t_max, a,K):
    t_values = np.arange(t0, t_max + h, h) 
    y_values = np.zeros(len(t_values))
    y_values[0] = y0

    for i in range(len(t_values) - 1):
        # équation à résoudre: y_values[i + 1] = y_values[i] + h * f(y_values[i+1], a)
        # y_values[i + 1] - y_values[i] - h * f(y_values[i], a) = 0
        # Z - y_values[i] - h * f(Z) = 0
        fonction_eq = lambda Z : Z - y_values[i] - h * verhulst(Z, a, K) 
        # On cherche la meilleure estimation de y[i + 1]; y[i] est la meilleure car le pas est faible et le valeurs ne change pas beaucoup
        point_initial = y_values[i] # fsolve résout par itération on définit ici le point de départ
        y_values[i + 1] = fsolve(fonction_eq, point_initial)
    return t_values, y_values
```

```python
def trapezes2_verhulst(y0, t0, h, t_max, a,K):
    t_values = np.arange(t0, t_max + h, h)
    y_values = np.zeros(len(t_values))
    y_values[0] = y0
    for i in range(len(t_values) - 1):
        euler_explicit = y_values[i] + h * verhulst(y_values[i], a,K)

        fonction_eq = lambda Z : Z - y_values[i] - h * verhulst(Z, a,K)
        # On cherche la meilleure estimation de y[i + 1]; y[i] est la meilleure car le pas est faible et le valeurs ne change pas beaucoup
        point_initial = y_values[i] # fsolve résout par itération on définit ici le point de départ
        euler_implicite = fsolve(fonction_eq, point_initial)

        y_values[i+1] = (euler_explicit + euler_implicite)/2
    return t_values, y_values
```

```python
def lotka_voltera_x(a, b, x, y):
    return a * x - b * x * y

def lotka_voltera_y(c, d, x, y):
    return -c * y + d * x * y
```

```python
def explicit_euler_method_lotka(y0, x0, t0, h, t_max, a, b, c, d):
    t_values = np.arange(t0, t_max + h, h)
    y_values = np.zeros(len(t_values))
    x_values = np.zeros(len(t_values))
    y_values[0] = y0
    x_values[0] = x0

    for i in range(len(t_values) - 1):
        x_temp = x_values[i]
        y_temp = y_values[i]
        y_values[i + 1] = y_values[i] + h * lotka_voltera_y(c, d, x_temp, y_temp)
        x_values[i + 1] = x_values[i] + h * lotka_voltera_x(a, b, x_temp, y_temp)

    return t_values, y_values, x_values
```

```python
y0 = 20      
x0 = 10  
t00 = 0
h0 = 0.01
t_max0 = 100
a1 = 0.5
b = 0.02
c = 0.3
d = 0.01

temps3, population_y, population_x = explicit_euler_method_lotka(y0, x0, t00, h0, t_max0, a1, b, c, d)

plt.plot(temps3, population_x, marker='o', color='tab:blue', linestyle='-', linewidth=2, label="Proies (x)")
plt.plot(temps3, population_y, marker='s', color='tab:orange', linestyle='-', linewidth=2, label="Prédateurs (y)")

plt.title("Évolution des populations - Méthode explicite d'Euler", fontsize=14)
plt.xlabel("Temps t", fontsize=12)
plt.ylabel("Taille de la population", fontsize=12)
plt.legend()
plt.show()
```
